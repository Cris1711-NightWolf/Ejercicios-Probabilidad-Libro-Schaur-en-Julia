```julia
# Ejercicio 4.94 — suma de variables Poisson independientes
# Si X₁ ~ Poisson(λ₁) y X₂ ~ Poisson(λ₂) son independientes:
# M_X₁(t)=exp(λ₁(e^t-1)) y M_X₂(t)=exp(λ₂(e^t-1)).
# Por independencia, M_(X₁+X₂)(t)=M_X₁(t)M_X₂(t)
# = exp((λ₁+λ₂)(e^t-1)), que es la MGF de Poisson(λ₁+λ₂).
# Para n variables independientes, el producto de las MGF da
# exp((λ₁+...+λₙ)(e^t-1)); por tanto la suma es Poisson(Σλᵢ).
using Distributions
λ1, λ2 = 2.0, 3.0
println("La suma X₁+X₂ tiene distribución Poisson(", λ1+λ2, ").")
println("P(X₁+X₂=4) = ", pdf(Poisson(λ1+λ2),4))
```
